# import libraries

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

# Data Reading


```python
spark.read.format('parquet')\
    .option('inferSchema',True)\
        .load('abfss://bronze@storagecarsud1.dfs.core.windows.net/rawdata').show()
```

```python
df=spark.read.format('parquet')\
    .option('inferSchema',True)\
        .load('abfss://bronze@storagecarsud1.dfs.core.windows.net/rawdata')
```

```python
df.display()
```

# Transformations

```python
df=df.withColumn('Model_Category',split('Model_ID','-')[0])
```

```python
display(df)
```

```python
df=df.withColumn('revenue_perUnit',col('Revenue')/col("Units_Sold"))
```

# Ad-HOC --pie chart

```python
df.groupBy("Year","BranchName").agg(sum("Units_Sold").alias('Total_units')).sort('Year','Total_units',ascending=[1,0]).display()
```

```python
df.write.format('parquet')\
    .mode('overwrite')\
        .option('path','abfss://silver@storagecarsud1.dfs.core.windows.net/cars_data_silver')\
            .save()
```
